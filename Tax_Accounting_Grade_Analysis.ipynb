```python
!pip install faker
import pandas as pd
from faker import Faker

sheet_id = "1kI5x_Tpw1UpNxoOeArQUalkIXWl7JY-23uWkurNldiA"
url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv"
df = pd.read_csv(url)

fake = Faker('ar_SA')
df['name'] = [fake.name() for _ in range(len(df))]
df = df[['name'] + [col for col in df.columns if col != 'name']]

df['letter_grade'] = pd.cut(
    df['final grade'],
    bins=[0, 10, 14, 18, 22, 30],
    labels=['F', 'D', 'C', 'B', 'A']
)

print("Grade distribution:")
print(df['letter_grade'].value_counts().sort_index())
print("\nClass statistics:")
print(df[['Q1', 'Q2', 'final grade']].describe())

df_clean = df.dropna(subset=['final grade'])
print(f"\nRows after dropping missing cells: {len(df_clean)}")

print(f"Total students: {len(df_clean)}")
print(f"Class average: {df_clean['final grade'].mean():.1f}/25")
print(f"Pass rate: {(df_clean['final grade'] >= 11).sum() / len(df_clean) * 100:.0f}%")
print(f"A students: {(df_clean['letter_grade'] == 'A').sum()}")
print(f"F students needing intervention: {(df_clean['letter_grade'] == 'F').sum()}")
```
